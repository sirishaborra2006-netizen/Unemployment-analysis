**DISTRUBUTION OF UNEMPLOYMENT RATE**

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.histplot(df['Estimated Unemployment Rate (%)'],
             bins=20,
             kde=True)

plt.title('Distribution of Unemployment Rate')
plt.show()
```

**TOP 10 STATES WITH HIGHEST UNEMPLOYEMENT**

```python
top_states = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean()

top_states = top_states.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_states.plot(kind='bar')

plt.title('Top 10 States by Average Unemployment Rate')
plt.ylabel('Unemployment Rate (%)')
plt.show()
```

**URBAN VS RURAL UNEMPLOYMENT**

```python
sns.boxplot(
    x='Area',
    y='Estimated Unemployment Rate (%)',
    data=df
)

plt.title('Urban vs Rural Unemployment')
plt.show()
```

```python
plt.figure(figsize=(12,5))

df.groupby('Date')['Estimated Unemployment Rate (%)'].mean().plot()

plt.title('Impact of COVID-19 on Unemployment')
plt.ylabel('Unemployment Rate (%)')
plt.xlabel('Date')

plt.show()
```

**Conclusion**

The analysis showed that unemployment rates varied across different states. Some states such as Tripura and Haryana had higher average unemployment rates compared to others.
*  The comparison between urban and rural areas showed that urban areas generally had higher unemployment rates than rural areas.
